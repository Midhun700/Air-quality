```python
import pandas as pd
import os

# Folder containing the dataset
dataset_folder = r"D:\Air Quality\dataset"

# Kerala station files
kerala_files = [
    "KL001.csv",
    "KL002.csv",
    "KL003.csv",
    "KL004.csv",
    "KL005.csv",
    "KL006.csv",
    "KL007.csv",
    "KL008.csv",
    "KL009.csv"
]

# Store individual station data
kerala_data = []

# Read each station file
for file in kerala_files:
    
    file_path = os.path.join(dataset_folder, file)
    
    df_station = pd.read_csv(file_path)
    
    # Add station ID
    df_station["Station_ID"] = file.replace(".csv", "")
    
    kerala_data.append(df_station)

# Combine all stations
kerala_df = pd.concat(kerala_data, ignore_index=True)

print("Kerala dataset created successfully!")
print("Shape:", kerala_df.shape)

display(kerala_df.head())
```

```python
# Select the station information we need
station_details = kerala_stations[
    ["file_name", "city", "station_location", "agency", "start_year"]
].copy()

# Rename file_name so it matches our dataset
station_details = station_details.rename(
    columns={"file_name": "Station_ID"}
)

# Merge station information into the Kerala air-quality data
kerala_df = kerala_df.merge(
    station_details,
    on="Station_ID",
    how="left"
)

print("Updated Kerala dataset:")
print("Shape:", kerala_df.shape)

display(kerala_df.head())
```

```python
output_path = r"D:\Air Quality\kerala_air_quality_raw.csv"

kerala_df.to_csv(output_path, index=False)

print("Saved successfully!")
print(output_path)
```

```python
# Check missing values in every column

missing = pd.DataFrame({
    "Missing_Count": kerala_df.isnull().sum(),
    "Missing_Percentage": (kerala_df.isnull().sum() / len(kerala_df)) * 100
})

missing = missing.sort_values(
    by="Missing_Percentage",
    ascending=False
)

display(missing)
```

```python
# Number of rows where ALL pollutant/weather measurements are missing

measurement_columns = [
    col for col in kerala_df.columns
    if col not in [
        "Station_ID",
        "city",
        "station_location",
        "agency",
        "start_year"
    ]
]

empty_rows = kerala_df[measurement_columns].isnull().all(axis=1)

print("Total rows:", len(kerala_df))
print("Completely empty measurement rows:", empty_rows.sum())
print(
    "Percentage:",
    round(empty_rows.mean() * 100, 2),
    "%"
)
```

```python
kerala_df["From Date"] = pd.to_datetime(
    kerala_df["From Date"],
    errors="coerce"
)

print("Earliest measurement:")
print(kerala_df["From Date"].min())

print("\nLatest measurement:")
print(kerala_df["From Date"].max())
```

```python
# Main pollutant columns for the project

pollutants = [
    "PM2.5 (ug/m3)",
    "PM10 (ug/m3)",
    "NO (ug/m3)",
    "NO2 (ug/m3)",
    "NOx (ppb)",
    "NH3 (ug/m3)",
    "SO2 (ug/m3)",
    "CO (mg/m3)",
    "Ozone (ug/m3)"
]

# Metadata columns
metadata = [
    "From Date",
    "To Date",
    "Station_ID",
    "city",
    "station_location",
    "agency",
    "start_year"
]

# Create our working dataset
air_df = kerala_df[metadata + pollutants].copy()

print("Working dataset shape:", air_df.shape)

display(air_df.head())
```

```python
missing_main = pd.DataFrame({
    "Missing_Count": air_df[pollutants].isnull().sum(),
    "Missing_Percentage": (
        air_df[pollutants].isnull().sum() / len(air_df)
    ) * 100
})

display(missing_main.sort_values(
    "Missing_Percentage",
    ascending=False
))
```

```python
before = len(air_df)

air_df = air_df.dropna(
    subset=pollutants,
    how="all"
).copy()

after = len(air_df)

print("Rows before:", before)
print("Rows after:", after)
print("Rows removed:", before - after)
```

```python
air_df["From Date"] = pd.to_datetime(
    air_df["From Date"],
    errors="coerce"
)

print("Date conversion completed.")
print("Start:", air_df["From Date"].min())
print("End:", air_df["From Date"].max())
```

```python
air_df["Date"] = air_df["From Date"].dt.date

print(air_df[["From Date", "Date"]].head())
```

```python
daily_pollutants = [
    "PM2.5 (ug/m3)",
    "PM10 (ug/m3)",
    "NO (ug/m3)",
    "NO2 (ug/m3)",
    "NOx (ppb)",
    "NH3 (ug/m3)",
    "SO2 (ug/m3)",
    "CO (mg/m3)",
    "Ozone (ug/m3)"
]

daily_df = (
    air_df
    .groupby(["Date", "Station_ID", "city"], as_index=False)[daily_pollutants]
    .mean()
)

print("Daily dataset created!")
print("Shape:", daily_df.shape)

display(daily_df.head())
```

```python
import matplotlib.pyplot as plt

daily_kerala = (
    daily_df
    .groupby("Date")["PM2.5 (ug/m3)"]
    .mean()
)

plt.figure(figsize=(14, 6))

plt.plot(
    daily_kerala.index,
    daily_kerala.values,
    linewidth=1
)

plt.xlabel("Date")
plt.ylabel("Average PM2.5 (µg/m³)")
plt.title("Daily Average PM2.5 in Kerala (2017–2023)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Make a copy
eda_df = daily_df.copy()

# Make sure Date is datetime
eda_df["Date"] = pd.to_datetime(eda_df["Date"], errors="coerce")

print("EDA dataset shape:", eda_df.shape)
print("\nDate range:")
print(eda_df["Date"].min(), "to", eda_df["Date"].max())

print("\nColumns:")
print(eda_df.columns.tolist())

display(eda_df.head())
```

```python
# Create Year and Month columns
eda_df["Year"] = eda_df["Date"].dt.year
eda_df["Month"] = eda_df["Date"].dt.month

# Monthly average PM2.5
monthly_pm25 = (
    eda_df
    .groupby(["Year", "Month"])["PM2.5 (ug/m3)"]
    .mean()
    .reset_index()
)

# Create proper date for plotting
monthly_pm25["Date"] = pd.to_datetime(
    monthly_pm25["Year"].astype(str) + "-" +
    monthly_pm25["Month"].astype(str) + "-01"
)

print("Monthly PM2.5 dataset:")
display(monthly_pm25.head())
```

```python
plt.figure(figsize=(15, 6))

plt.plot(
    monthly_pm25["Date"],
    monthly_pm25["PM2.5 (ug/m3)"],
    linewidth=1.5
)

plt.title("Monthly Average PM2.5 in Kerala (2017–2023)")
plt.xlabel("Date")
plt.ylabel("PM2.5 (µg/m³)")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
yearly_pm25 = (
    eda_df
    .groupby("Year")["PM2.5 (ug/m3)"]
    .mean()
    .reset_index()
)

print("Year-wise average PM2.5:")
display(yearly_pm25)
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    yearly_pm25["Year"].astype(str),
    yearly_pm25["PM2.5 (ug/m3)"]
)

plt.title("Year-wise Average PM2.5 in Kerala")
plt.xlabel("Year")
plt.ylabel("Average PM2.5 (µg/m³)")

plt.tight_layout()
plt.show()
```

```python
pollutants = [
    "PM2.5 (ug/m3)",
    "PM10 (ug/m3)",
    "NO (ug/m3)",
    "NO2 (ug/m3)",
    "NOx (ppb)",
    "NH3 (ug/m3)",
    "SO2 (ug/m3)",
    "CO (mg/m3)",
    "Ozone (ug/m3)"
]

pollutant_means = eda_df[pollutants].mean().sort_values(ascending=False)

print("Average pollutant concentrations:")
display(pollutant_means)
```

```python
def assign_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Summer"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Post-Monsoon"

eda_df["Season"] = eda_df["Date"].dt.month.map(assign_season)

print(eda_df[["Date", "Season"]].head(10))
```

```python
season_pm25 = (
    eda_df
    .groupby("Season")["PM2.5 (ug/m3)"]
    .mean()
)

print("Season-wise average PM2.5:")
display(season_pm25)
```

```python
season_order = [
    "Winter",
    "Summer",
    "Monsoon",
    "Post-Monsoon"
]

season_pm25 = season_pm25.reindex(season_order)

plt.figure(figsize=(9, 5))

season_pm25.plot(kind="bar")

plt.title("Seasonal Variation of PM2.5 in Kerala")
plt.xlabel("Season")
plt.ylabel("Average PM2.5 (µg/m³)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
